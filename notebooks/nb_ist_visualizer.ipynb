```python
import pandas as pd 
import seaborn as sns
```

```python
# chose your path for the df_trg.csv file
filepath = "../data/builds/dyconex_251117/control/df_trg.csv"

df = pd.read_csv(filepath)
df.head()
```

## Important columns
- `group`: this is an ID used to query the respective process data.
- `position`: the x-axis of the IST, i.e. the cycle number of the thermal loading test
- `variable`: each IST test structure has two circuits to test, they are Sense A and Sense B. They belong to the same device but have slightly different via design.
- `value`: this is the percent resistance increase from the initial resistance at position = 0. The value is already normalized for the ML method, to get back to the non-normalized value, just multiply by 10. 

```python
groups = df["group"].unique()

# plot the first 5 groups, plotting all of them takes some time
sns.lineplot(data=df.set_index("group").loc[groups[:5]], x="position", y="value", hue="group", style="variable")
```
